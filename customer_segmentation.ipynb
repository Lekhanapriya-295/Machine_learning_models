```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("store_customers.csv")
print(data.head())
data = data.drop("CustomerID",axis=1)
```

```python
data.shape
```

```python
sns.histplot(data['Age'],color='green')
plt.show()
```

```python
sns.histplot(data['Annual Income (k$)'])
plt.show()
```

```python
sns.histplot(data['Spending Score (1-100)'],color='yellow')
plt.show()
```

```python
sns.scatterplot(x=data['Annual Income (k$)'],y=data['Spending Score (1-100)'])
plt.show()
```

```python
data.dropna(inplace=True)
x = data[['Annual Income (k$)','Spending Score (1-100)']]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
kmeans = KMeans(n_clusters=5)
clusters = kmeans.fit_predict(x_scaled)
data['Cluster'] = clusters
```

```python
plt.scatter(data['Annual Income (k$)'],data['Spending Score (1-100)'],c=data['Cluster'])
plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.show()
```
